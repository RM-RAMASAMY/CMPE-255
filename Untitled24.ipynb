```python
import streamlit as st
import pandas as pd
from surprise import SVD, Reader, Dataset
from surprise.model_selection import train_test_split
from surprise import accuracy
import numpy as np

# Step 1: Business Understanding
# Problem Statement:
# This project aims to develop a movie recommendation system that helps users discover movies based on their preferences.
# Success Metrics:
# - Accuracy: Precision and recall should be above 80%.
# - User Engagement: 20% increase in movie interactions.
# - Diversity: 50% of recommendations should be unseen by the user.

# Step 2: Data Understanding
# Load the MovieLens dataset (change the path to your local file paths)
@st.cache
def load_data():
    # Update with your local file paths
    movies = pd.read_csv("movies[1].csv")
    ratings = pd.read_csv("ratings[1].csv")
    return movies, ratings

movies, ratings = load_data()

# Display dataset information
st.title("Movie Recommendation System")
st.write("Welcome to the Movie Recommendation System! This system recommends movies based on your preferences.")
st.write("### Data Preview:")
st.write("#### Movies Dataset:")
st.write(movies.head())

st.write("#### Ratings Dataset:")
st.write(ratings.head())

# Step 3: Data Preparation
# Create a Surprise reader object to define the dataset format
reader = Reader(rating_scale=(0.5, 5.0))

# Prepare the data for Surprise library (Dataset class)
data = Dataset.load_from_df(ratings[['userId', 'movieId', 'rating']], reader)

# Split the dataset into training and testing
trainset, testset = train_test_split(data, test_size=0.2)

# Step 4: Modeling
# Use the SVD (Singular Value Decomposition) algorithm for collaborative filtering
model = SVD()

# Train the model on the training set
model.fit(trainset)

# Step 5: Evaluation
# Evaluate the model using RMSE (Root Mean Square Error)
predictions = model.test(testset)
rmse = accuracy.rmse(predictions)

# Display RMSE
st.write(f"### Model Evaluation:")
st.write(f"Root Mean Square Error (RMSE) on test set: {rmse:.4f}")

# Function to get movie recommendations
def get_movie_recommendations(user_id, top_n=10):
    # Get the list of all movie IDs
    movie_ids = movies['movieId'].tolist()

    # Predict ratings for all movies for the given user
    predictions = [model.predict(user_id, movie_id) for movie_id in movie_ids]

    # Sort predictions based on estimated ratings (descending)
    predictions.sort(key=lambda x: x.est, reverse=True)

    # Get the top N movie recommendations
    top_recommendations = predictions[:top_n]

    recommended_movie_ids = [pred.iid for pred in top_recommendations]

    # Get movie names for the recommended movie IDs
    recommended_movies = movies[movies['movieId'].isin(recommended_movie_ids)]

    return recommended_movies

# Step 6: Deployment (Streamlit UI)
st.write("### Movie Recommendation")

# Allow the user to input their User ID for personalized recommendations
user_id = st.number_input("Enter your User ID for personalized recommendations", min_value=1, max_value=610, value=1)

# Get top N movie recommendations for the given user
top_n = 10
recommended_movies = get_movie_recommendations(user_id, top_n)

st.write(f"#### Top {top_n} Recommended Movies for User {user_id}:")
st.write(recommended_movies[['movieId', 'title', 'genres']])

# Allow the user to click and get more movie details
selected_movie = st.selectbox("Select a movie for more details:", recommended_movies['title'].tolist())

if selected_movie:
    movie_details = recommended_movies[recommended_movies['title'] == selected_movie].iloc[0]
    st.write(f"### Movie Details for {selected_movie}")
    st.write(f"*Movie ID:* {movie_details['movieId']}")
    st.write(f"*Genres:* {movie_details['genres']}")

# Optional: Show the full dataset
if st.checkbox("Show full Movies dataset"):
    st.write(movies)

# Optional: Show the full Ratings dataset
if st.checkbox("Show full Ratings dataset"):
    st.write(ratings)
```
