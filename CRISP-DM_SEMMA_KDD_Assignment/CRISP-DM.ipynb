# Exploratory Data Analysis (EDA) Using CRISP-DM Methodology

## Business Understanding

The goal of this analysis is to understand customer segments and their purchasing patterns within a mall. This will help in developing targeted marketing strategies and improving customer experience.

## Data Understanding

The dataset contains 200 records with the following columns:
- **CustomerID**: Unique identifier for each customer
- **Genre**: Customer's gender (Male/Female)
- **Age**: Customer's age
- **Annual Income (k$)**: Customer's yearly income in thousands of dollars
- **Spending Score (1-100)**: Score assigned by the mall based on customer behavior and spending nature

## Data Preparation

The data is clean with no missing values. No significant preprocessing is required.

## Data Analysis

### Descriptive Statistics

- **Age**: 18 to 70 years (mean: 38.85)
- **Annual Income**: $15,000 to $137,000 (mean: $60,560)
- **Spending Score**: 1 to 99 (mean: 50.2)

### Age Distribution



Key observations:
- Right-skewed distribution with most customers between 30 and 50 years old.
- Peak around age 32-35.
- Fewer customers in older age ranges (60+).

### Income vs. Spending Score

Scatter plot analysis shows:
- No strong linear correlation between income and spending score.
- Potential clusters of customers with different income and spending patterns.

## Evaluation

Insights from the EDA:
1. The mall attracts a diverse age group, with a concentration of customers in their 30s and 40s.
2. There's a wide range of annual incomes and spending scores, suggesting diverse customer segments.
3. The relationship between income and spending is not straightforward, indicating that factors other than income influence spending behavior.

## Deployment

Recommendations:
1. Develop targeted marketing strategies for different age groups, especially focusing on the 30-50 age range.
2. Investigate factors beyond income that influence spending scores to better understand customer behavior.
3. Consider creating customer segments based on observed clusters for personalized marketing approaches.

Further analysis could include clustering algorithms or more detailed demographic breakdowns for additional insights.
